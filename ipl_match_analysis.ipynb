### head() method displays the first 5 rows by default.
#### We can pass an integer argument to specify the number of rows to display.


```
data.head() # displays the first 5 rows
```

```
data.head(10) # displays the first 10 rows
```

### tail() method displays the last 5 rows by default.
#### We can pass an integer argument to specify the number of rows to display.

```
data.tail() # displays the last 5 rows
```

```
data.tail(3) # displays the last 3 rows
```
