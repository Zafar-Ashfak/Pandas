### Write an operation to find the top 3 batsmen with the highest total runs scored.

```
import pandas as pd
```

```
data = pd.read_csv('deliveries.csv')
data
```

```
all_batsman = data.groupby('batsman')
```

```
all_batsman
```

```
all_batsman['batsman_runs'].sum().sort_values(ascending=False).head(3)
```
