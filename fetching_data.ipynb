# Fetching rows and cols

```
import numpy as np
import pandas as pd
```

```
data = pd.read_csv('matches.csv')
```

### Checking datatype of single col

```
print(type(data['winner'])) # datatype -> Series
```

### Fetching single column data

```
data['winner']
```

### Checking datatype of multiple cols

```
print(type(data[['team1', 'team2', 'winner']])) # datatype -> DataFrame
```

### Fetching multiple cols data

```
data[['team1', 'team2', 'winner']]
```

### Fetching rows using iloc property

```
data.iloc[0] # fetching data of 0th row
```

```
data.iloc[4] # fetching data of 4th row
```

### Fetching multiple row using iloc property with slicing

```
data.iloc[3 : 6] # 3 -> 5
```

```
data.iloc[ : 9] # 0 -> 8
```

```
data.iloc[0 : 21 : 2] # step-wise fetching
```

### Fetching rows using numpy indexing

```
data.iloc[[1, 2, 5, 9, 11, 12, 18]] # indexes
```

### Fetching cols using iloc property

```
data.iloc[:, [4, 5, 10]] # # fetching data of 4th, 5th and 10th col
```

## Filtering data using mask

```
mask = data['city'] == 'Hyderabad'
print(type(mask)) # type of mask data is boolean series

data[mask]
```

```
data[mask].shape[0] # displays the winner city count
```

### Q. Write a function to count the winner city

```
def get_winner_city_count(wining_city):
    winning_mask = data['city'] == wining_city
    return data[winning_mask].shape[0]
```

```
# get_winner_city_count('Hyderabad') # 49
get_winner_city_count('Rajkot') # 10
```

```
mask1 = data['winner'] == 'Sunrisers Hyderabad'
mask2 = data['season'] > 2012


print(f"Sunrisers Hyderabad won {data[mask1 & mask2].shape[0]} times after 2012") # displays to winner city after 2012
data.iloc[mask1 & mask2]

```
