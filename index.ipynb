# set_index() and reset_index()

```
import pandas as pd
```

```
df = pd.read_csv('matches.csv')
df
```

### set_index()
###### This method is used to turn one or more existing columns into the row index (row labels) of a DataFrame

```
df.set_index('id', drop=True)
# df.set_index('id', drop=True, inplace=True)          To make permanent changes.
```

### reset_index(drop=True)
###### This method resets a DataFrame's row index to the default 0, 1, 2... integer sequence and deletes the old index entirely

```
df.reset_index(drop=True)
```

### Another use of reset_index()

```
wt = df['winner'].value_counts().sort_values(ascending=False)
print(type(wt))
```

```
wt.reset_index()
```
