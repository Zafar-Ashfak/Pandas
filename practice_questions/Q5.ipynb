### Write an operation to find the number of sixes hit in each over.

```
import pandas as pd
import seaborn as sb
```

```
df = pd.read_csv('../deliveries.csv')
df.head(3)
```

```
mask = df['batsman_runs'] == 6
sixes = df[mask]
sixes
```

```
pt = sixes.pivot_table(index='over', columns='batting_team', values='batsman_runs', aggfunc='count')
```

```
sb.heatmap(pt)
```
