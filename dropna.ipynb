# dropna()

```
import pandas as pd
```

```
df = pd.read_csv('train.csv')
print(df.shape)
df
```

### syntax -> df.dropna(axis=0, how='any', thresh=None, subset=None, inplace=False, ignore_index=False)

### 1. Drop rows with any missing values (Default Behavior)

```
dropped_any = df.dropna()
# dropped_any = df.dropna(how='any', inplace=True)
print(dropped_any.shape)
dropped_any
```

### 2. Drop rows only if all values are missing

```
dropped_all = df.dropna(how='all')
print(dropped_all.shape)
dropped_all
```
