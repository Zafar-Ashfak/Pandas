### Write an operation to show the orange cap holder of each season in IPL

```
import pandas as pd
```

```
matches = pd.read_csv('../matches.csv')
deliveries = pd.read_csv('../deliveries.csv')
```

#### Merging the matches and deliveries DataFrame to get the orange cap holders

```
new_df = deliveries.merge(matches, left_on='match_id', right_on='id')
print(deliveries.shape)
print(matches.shape)
print(new_df.shape)
```

```
all_batsman = new_df.groupby(['season', 'batsman'])['batsman_runs'].sum().sort_values(ascending=False)
```

```
sorted_df = all_batsman.reset_index().drop_duplicates('season', keep='first').sort_values('season')
sorted_df
```

### Showing season and its batsman

```
sorted_df[['season', 'batsman']]
```
