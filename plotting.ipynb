# Graph plotting

```
import pandas as pd
import matplotlib.pyplot as plt
```

```
data = pd.read_csv('matches.csv')

data.iloc[0]
```

### Bar graph of winner team

```
data['winner'].value_counts().plot(kind='bar')
```

### Vertical Bar graph of top 3 winning team

```
data['winner'].value_counts().head(3).plot(kind='bar')
```

### Horizontal bar graph of top 5 winning team

```
data['winner'].value_counts().head(5).plot(kind='barh')
```

### Plotting Pie chart on toss decision

```
data['toss_decision'].value_counts().plot(kind='pie')
```

### Plotting histogram on win by runs

```
data['win_by_runs'].plot(kind='hist')
```
