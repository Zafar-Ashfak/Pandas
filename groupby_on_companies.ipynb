# Group By Operation on Fortune 500 Companies

```
import pandas as pd
```

```
companies = pd.read_excel('companies.xlsx')
print(f"The total number of rows in the DataFrame is: {companies.shape[0]}")
print(f"The total number of cols in the DataFrame is: {companies.shape[1]}")
```

```
companies
```

### Getting the information about the companies data

```
companies.info()
```

### Group by operation

```
sectors = companies.groupby('Sector')
```

```
print(f"The number of companies in the DataFrame is: {len(companies)}")
print(f"After dividing companies in the sectors is: {len(sectors)}")
```

```
sectors
```

### Find the size of the sectors from companies DataFrame

```
sectors.size()
```

```
sectors.size().sort_values(ascending=False)
```

### first()
This method will return the first occurrence of the matching sector

```
sectors.first()
```

### last()
This method will return the last occurrence of the matching sector

```
sectors.last()
```

### groups
This attribute returns the list of index of the groupby sectors

```
sectors.groups
```

### To verify if the groups attribute has returned the right indices

```
companies.iloc[[8, 13, 26, 30]]
```

### get_group()
This method returns the DataFrame of the particular group

```
sectors.get_group('Health Care')
```

### sum()
This method returns the sum of a DataFrame of all numerical cols

```
sectors.sum()
```
